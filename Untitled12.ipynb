```python
%pip install -q kagglehub pandas numpy matplotlib scikit-learn

import kagglehub
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import glob
import csv
import io
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

path = kagglehub.dataset_download("ankitgaikar1995/imd-rainfall-dataset-2022")
print("Dataset path:", path)

files_found = (
    glob.glob(os.path.join(path, "**", "*.csv"), recursive=True)
    + glob.glob(os.path.join(path, "**", "*.txt"), recursive=True)
)

if not files_found:
    raise FileNotFoundError("No CSV or TXT files found in the dataset.")

raw = None
file_path = None

for candidate in files_found:
    try:
        with open(candidate, "r", encoding="utf-8-sig", errors="replace") as f:
            sample = f.read(10000)

        if not sample.strip():
            continue

        try:
            delimiter = csv.Sniffer().sniff(
                sample[:5000], delimiters=",;\t|"
            ).delimiter
        except csv.Error:
            delimiter = ";"

        temp = pd.read_csv(
            candidate,
            sep=delimiter,
            encoding="utf-8-sig",
            engine="python",
            on_bad_lines="skip"
        )

        if temp.shape[1] == 1 and ";" in str(temp.columns[0]):
            temp = pd.read_csv(
                candidate,
                sep=";",
                encoding="utf-8-sig",
                engine="python",
                on_bad_lines="skip"
            )

        normalized = [
            str(c).strip().strip('"').lower()
            for c in temp.columns
        ]

        if "district" in normalized and "month" in normalized:
            raw = temp
            file_path = candidate
            break

        if any(c in normalized for c in ["date", "datetime"]) and any(
            c in normalized for c in ["rainfall", "rainfall_mm", "rain"]
        ):
            raw = temp
            file_path = candidate
            break

    except Exception as e:
        print("Skipping unreadable file:", os.path.basename(candidate), str(e))

if raw is None:
    raise ValueError("Could not locate a readable rainfall dataset.")

raw.columns = [
    str(c).strip().strip('"').lower()
    for c in raw.columns
]

print("Loaded file:", file_path)
print("Columns:", raw.columns.tolist())
print("Rows:", len(raw))

raw = raw.loc[:, ~raw.columns.duplicated()]

if {"district", "month"}.issubset(raw.columns):
    day_columns = [
        c for c in raw.columns
        if str(c).lower().strip().rstrip("stndrh") in [str(i) for i in range(1, 32)]
        or str(c).lower().strip() in [
            f"{i}st" for i in range(1, 32)
        ]
        or str(c).lower().strip() in [
            f"{i}nd" for i in range(1, 32)
        ]
        or str(c).lower().strip() in [
            f"{i}rd" for i in range(1, 32)
        ]
        or str(c).lower().strip() in [
            f"{i}th" for i in range(1, 32)
        ]
    ]

    def get_day_number(column):
        digits = "".join(ch for ch in str(column) if ch.isdigit())
        return int(digits) if digits else 0

    day_columns = sorted(set(day_columns), key=get_day_number)

    if not day_columns:
        raise ValueError("District/month format found, but daily rainfall columns are missing.")

    raw["month"] = pd.to_numeric(raw["month"], errors="coerce")
    raw = raw[raw["month"].between(1, 12)].copy()

    for c in day_columns:
        raw[c] = pd.to_numeric(raw[c], errors="coerce")
        raw.loc[raw[c] <= -999, c] = np.nan
        raw.loc[raw[c] < 0, c] = np.nan

    raw["district"] = raw["district"].astype(str).str.strip()
    raw["state"] = raw["state"].astype(str).str.strip() if "state" in raw.columns else "Unknown"

    locations = (
        raw[["state", "district"]]
        .drop_duplicates()
        .sort_values(["state", "district"])
        .reset_index(drop=True)
    )

    print("\nAvailable locations:")
    print(locations.to_string(index=False))

    selected_state = None
    selected_district = None

    if len(locations) > 1:
        selected_state = input("Enter state name from the list above: ").strip()
        state_rows = locations[
            locations["state"].str.casefold() == selected_state.casefold()
        ]

        if state_rows.empty:
            raise ValueError("State not found. Check the displayed location list.")

        selected_district = input("Enter district name from the selected state: ").strip()

        raw = raw[
            (raw["state"].str.casefold() == selected_state.casefold())
            & (raw["district"].str.casefold() == selected_district.casefold())
        ].copy()

        if raw.empty:
            raise ValueError("District not found in the selected state.")
    else:
        selected_state = locations.iloc[0]["state"]
        selected_district = locations.iloc[0]["district"]

    month_names = {
        "jan": 1, "january": 1,
        "feb": 2, "february": 2,
        "mar": 3, "march": 3,
        "apr": 4, "april": 4,
        "may": 5,
        "jun": 6, "june": 6,
        "jul": 7, "july": 7,
        "aug": 8, "august": 8,
        "sep": 9, "sept": 9, "september": 9,
        "oct": 10, "october": 10,
        "nov": 11, "november": 11,
        "dec": 12, "december": 12
    }

    if raw["month"].dtype == object:
        raw["month"] = raw["month"].astype(str).str.lower().str.strip().map(
            lambda x: month_names.get(x, pd.to_numeric(x, errors="coerce"))
        )

    records = []

    for _, row in raw.iterrows():
        month_value = pd.to_numeric(row["month"], errors="coerce")

        if pd.isna(month_value) or not 1 <= month_value <= 12:
            continue

        month_value = int(month_value)

        for day_col in day_columns:
            day_value = get_day_number(day_col)

            if day_value < 1 or day_value > 31:
                continue

            try:
                date_value = pd.Timestamp(2022, month_value, day_value)
            except ValueError:
                continue

            records.append({
                "Date": date_value,
                "Rainfall": row[day_col]
            })

    df = pd.DataFrame(records)

else:
    date_col = next(
        (c for c in raw.columns if c in ["date", "datetime", "observation_date"]),
        None
    )
    rain_col = next(
        (c for c in raw.columns if c in [
            "rainfall", "rainfall_mm", "rain", "precipitation", "rf"
        ]),
        None
    )

    if date_col is None or rain_col is None:
        raise ValueError("Unrecognized dataset structure.")

    df = raw[[date_col, rain_col]].copy()
    df.columns = ["Date", "Rainfall"]
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce", dayfirst=True)
    df["Rainfall"] = pd.to_numeric(df["Rainfall"], errors="coerce")
    df = df.dropna(subset=["Date"])
    df = df[df["Date"].dt.year == 2022]

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Rainfall"] = pd.to_numeric(df["Rainfall"], errors="coerce")
df = df.dropna(subset=["Date"])
df.loc[df["Rainfall"] <= -999, "Rainfall"] = np.nan
df.loc[df["Rainfall"] < 0, "Rainfall"] = np.nan
df = df[df["Date"].dt.year == 2022]

if df.empty:
    raise ValueError("No rainfall observations could be extracted for 2022.")

df = df.groupby("Date")["Rainfall"].mean().sort_index()

calendar = pd.date_range("2022-01-01", "2022-12-31", freq="D")
rain = df.reindex(calendar)
rain.index.name = "Date"

print("\nMissing daily observations before imputation:", int(rain.isna().sum()))

rain = rain.interpolate(method="time", limit=2, limit_area="inside")

monthly_medians = rain.groupby(rain.index.month).median()
overall_median = rain.median()

if pd.isna(overall_median):
    raise ValueError("All rainfall observations are missing or invalid.")

for month in range(1, 13):
    mask = (rain.index.month == month) & rain.isna()
    median_value = monthly_medians.get(month, np.nan)
    if pd.isna(median_value):
        median_value = overall_median
    rain.loc[mask] = median_value

rain = rain.clip(lower=0).astype(float)

print("Missing values after imputation:", int(rain.isna().sum()))
print("Daily observations:", len(rain))

plt.figure(figsize=(14, 5))
plt.plot(rain.index, rain.values, color="royalblue")
plt.title(f"Daily Rainfall Time Series - 2022 ({selected_district if 'selected_district' in locals() else 'Selected location'})")
plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

def create_features(series):
    features = pd.DataFrame(index=series.index)

    for lag in range(1, 8):
        features[f"lag_{lag}"] = series.shift(lag)

    previous = series.shift(1)
    features["mean_3_days"] = previous.rolling(3).mean()
    features["mean_7_days"] = previous.rolling(7).mean()
    features["max_7_days"] = previous.rolling(7).max()
    features["rainy_days_7"] = (previous >= 0.1).rolling(7).sum()
    features["month"] = features.index.month
    features["day_sin"] = np.sin(2 * np.pi * features.index.dayofyear / 365.25)
    features["day_cos"] = np.cos(2 * np.pi * features.index.dayofyear / 365.25)

    return features

X_all = create_features(rain)
dataset = X_all.copy()
dataset["target"] = rain
dataset = dataset.dropna()

X = dataset.drop(columns="target")
y = dataset["target"]

if len(dataset) < 60:
    raise ValueError("Insufficient data to train the model.")

split = int(len(dataset) * 0.8)

X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

model = RandomForestRegressor(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=3,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
test_predictions = np.maximum(model.predict(X_test), 0)

print("\nModel evaluation")
print("MAE:", round(mean_absolute_error(y_test, test_predictions), 3), "mm")
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, test_predictions)), 3), "mm")
print("R2:", round(r2_score(y_test, test_predictions), 3))
print("Baseline MAE:", round(mean_absolute_error(y_test, X_test["lag_1"]), 3), "mm")

plt.figure(figsize=(14, 5))
plt.plot(y_test.index, y_test.values, label="Actual rainfall")
plt.plot(y_test.index, test_predictions, label="Predicted rainfall")
plt.title("Actual vs Predicted Rainfall")
plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

final_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=3,
    random_state=42,
    n_jobs=-1
)
final_model.fit(X, y)

history = rain.copy()
forecast_rows = []
rain_threshold = 0.1
last_date = history.index.max()

for step in range(1, 8):
    forecast_date = last_date + pd.Timedelta(days=step)
    history.loc[forecast_date] = np.nan
    next_features = create_features(history).loc[[forecast_date]]
    prediction = max(0.0, float(final_model.predict(next_features)[0]))
    history.loc[forecast_date] = prediction

    forecast_rows.append({
        "Date": forecast_date,
        "Predicted_Rainfall_mm": round(prediction, 2),
        "Rain_Status": "Rain" if prediction >= rain_threshold else "No rain"
    })

forecast = pd.DataFrame(forecast_rows)

print("\nSeven-day rainfall forecast")
print(forecast.to_string(index=False))

rainy_days = forecast[forecast["Predicted_Rainfall_mm"] >= rain_threshold]

print("\nNext predicted rainy day")

if not rainy_days.empty:
    next_rain = rainy_days.iloc[0]
    print("Date:", next_rain["Date"].strftime("%Y-%m-%d"))
    print("Predicted rainfall:", next_rain["Predicted_Rainfall_mm"], "mm")
else:
    print("No rainy day predicted in the next seven days.")

plt.figure(figsize=(12, 5))
colors = [
    "royalblue" if value >= rain_threshold else "lightgray"
    for value in forecast["Predicted_Rainfall_mm"]
]

plt.bar(
    forecast["Date"].dt.strftime("%d %b"),
    forecast["Predicted_Rainfall_mm"],
    color=colors
)
plt.axhline(rain_threshold, color="red", linestyle="--", label="Rain threshold")
plt.title("Seven-Day Rainfall Forecast")
plt.xlabel("Date")
plt.ylabel("Predicted Rainfall (mm)")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

forecast.to_csv("IMD_7_Day_Rainfall_Forecast.csv", index=False)
print("Forecast saved to IMD_7_Day_Rainfall_Forecast.csv")
```
